### 2. Create a list of numbers and print only even numbers using for loop

```python
random_numbers=range(1,21)
even_numbers=[]
for i in random_numbers:
    if i%2==0:
        even_numbers.append(i)
print("Even Numbers:",even_numbers)
```

### 3.Write a program that handles division by zero using try-except

```python
First_num=int(input("Enter Your 1st Number:"))
Second_num=int(input("Enter Your 2nd Number:"))

try:
    Divide=First_num/Second_num
    print(Divide)
except ZeroDivisionError:
    print("Invalid Input. Enter Other Than 0")
finally:
    print("Completed")
```

### 4. Given a string, count how many vowels it has.

```python
user_input=input("Your Name:")
count=0
for i in user_input.lower():
    if i in "aeiou":
        count +=1
print(count)
    
```

### 5. reverse a list without using built-in functions.

```python
user_input=[1,2,3,4,5]
output=[]
for i in user_input:
    output.append(user_input[-1] - i+1)
print(output)
```

```python
original_list=[1,2,3,4,5]
reverse_list=[]
i=len(original_list)
while i>0:
    reverse_list.append(i)
    i-=1
print(reverse_list)
```
